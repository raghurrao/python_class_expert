## 3. Concept Explanation
PyTorch is an open-source machine learning framework developed by Meta AI. At its core, it is two things:
1. A replacement for NumPy to use the power of GPUs.
2. A deep learning research platform that provides maximum flexibility and speed.

A **Tensor** is the fundamental data structure in PyTorch. It is essentially a multi-dimensional array, very similar to a NumPy `ndarray`, but with superpowers (like running on a GPU and tracking gradients).

```
import torch
import numpy as np

print(f"PyTorch Version: {torch.__version__}")
```

## 4. Why This Matters
In traditional programming, you write rules. In ML, the model learns the rules from data. To process massive amounts of data, we need specialized data structures (tensors) that can run on specialized hardware (GPUs). NumPy is great, but it's bound to the CPU. PyTorch bridges this gap.

## 5. Intuition
Think of a tensor as a spreadsheet:
- 0D Tensor (Scalar): A single cell with a number.
- 1D Tensor (Vector): A single column of numbers.
- 2D Tensor (Matrix): A sheet with rows and columns.
- 3D Tensor: A workbook with multiple sheets.

## 6. Mathematical Foundation
Tensors are mathematical entities. A 1D tensor $\vec{v} \in \mathbb{R}^n$ represents a point in $n$-dimensional space. We use them to represent data: for instance, a sequence of words, the pixels of an image, or user features.

## 7. PyTorch API
We use `torch.tensor(data)` to create a tensor from Python lists or NumPy arrays.

```
# 8. Simple Example
data = [1, 2, 3, 4]
my_first_tensor = torch.tensor(data)
print("Tensor:", my_first_tensor)
print("Type:", type(my_first_tensor))
```

## 9. Code Walkthrough
- `data = [1, 2, 3, 4]`: We start with a standard Python list.
- `torch.tensor(data)`: This factory function copies the data and constructs a PyTorch tensor.
- `print(...)`: Notice the output shows `tensor([1, 2, 3, 4])` indicating its PyTorch nature.

## 10. Experiment
