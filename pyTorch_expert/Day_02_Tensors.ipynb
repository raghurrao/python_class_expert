## 3. Concept Explanation
While we can create tensors from Python lists, in practice, we usually initialize them with specific patterns (zeros, ones, random numbers) or create them directly on the GPU. 

Every tensor has 3 key attributes you must ALWAYS know:
1. **`shape`**: The size of each dimension.
2. **`dtype`**: The data type of the elements (e.g., float32, int64).
3. **`device`**: Where the tensor lives (CPU or GPU).

## 4. Why This Matters
Deep learning is entirely about matrix multiplications and transformations. If your shapes don't match, your code crashes. If your `dtype` is wrong, PyTorch will throw an error. If one tensor is on the CPU and another is on the GPU, they cannot interact. Mastering these fundamentals saves you hundreds of hours of debugging.

## 5. Intuition
Think of **Reshaping** like pouring water from a wide, shallow glass into a tall, narrow one. The total amount of water (total number of elements) stays the same, but the container's shape changes.

Think of **Squeezing** as removing useless outer packaging. If you have a box containing exactly 1 smaller box, which contains your item, you can throw away the outer box. A tensor of shape `[1, 5, 1]` can be squeezed to `[5]`.

## 7. PyTorch API - Tensor Creation
Here are the most common ways to create tensors from scratch.

```
# 8. Simple Example: Creation
zeros = torch.zeros(3, 4)      # 3 rows, 4 columns of zeros
ones = torch.ones(2, 2)        # 2x2 matrix of ones
rand_tensor = torch.rand(2, 3) # Random values from uniform distribution [0, 1)
arange_tensor = torch.arange(0, 10, step=2) # [0, 2, 4, 6, 8]

print("Random Tensor:\n", rand_tensor)
```

## 9. Code Walkthrough: Attributes
Let's inspect a tensor's attributes.

```
example = torch.rand(3, 4)

print(f"Shape: {example.shape}")
print(f"Datatype: {example.dtype}")
print(f"Device: {example.device}")
```

## 10. Experiment: Indexing and Slicing
Indexing works exactly like NumPy or Python lists.

```
tensor = torch.arange(1, 10).reshape(3, 3)
print("Original:\n", tensor)

print("\nFirst row:", tensor[0])
print("First column:", tensor[:, 0])
print("Last element:", tensor[-1, -1])
```

## 12. Practice Exercise 2: Squeeze and Unsqueeze
Sometimes models expect a batch dimension. A single image of shape `[28, 28]` might need to become `[1, 28, 28]`.
`unsqueeze(dim)` adds a dimension of size 1 at the specified index.
`squeeze()` removes all dimensions of size 1.

**Task:** Create a tensor of shape `(5, 5)`. Add a batch dimension at the beginning so the shape becomes `(1, 5, 5)`. Then remove it.

```
# Write your code here
```

```
# SOLUTION
base = torch.ones(5, 5)
unsqueezed = base.unsqueeze(0)
print("Unsqueezed shape:", unsqueezed.shape)

squeezed = unsqueezed.squeeze(0)
print("Squeezed shape:", squeezed.shape)
```

## 13. Debugging Challenge
You want to reshape a tensor of size 10 into a 3x3 matrix. Why does this fail? How do you fix the code to make it work (e.g., by changing the initial tensor size)?

```
a = torch.arange(10)
# b = a.reshape(3, 3) # Uncomment to see the error
```

**Solution:** A 3x3 matrix requires exactly 9 elements (3 * 3 = 9). The tensor `a` has 10 elements. You cannot reshape 10 elements into 9 slots. You must either slice the tensor to have 9 elements `a[:9].reshape(3, 3)` or create `torch.arange(9)`.

## 16. Common Mistakes
- Confusing `view()` and `reshape()`. `view()` is faster but requires the tensor memory to be contiguous. `reshape()` handles non-contiguous memory safely by copying if necessary. When in doubt, use `reshape()`.
- Hardcoding dimensions instead of using `-1`. In `reshape()`, you can use `-1` for one dimension to let PyTorch infer it. E.g., `tensor.reshape(2, -1)` will automatically figure out the second dimension size.

## 17. Interview Questions
1. **What is the difference between `view` and `reshape` in PyTorch?**
   *Answer*: `view` operates on the existing memory block and requires it to be contiguous. `reshape` can operate on non-contiguous tensors by copying the data to a new memory block if needed.
2. **How do you add a dummy batch dimension to a single image tensor?**
   *Answer*: Use `image.unsqueeze(0)` to add a dimension at index 0.
