## 3. Concept Explanation
In Day 6, we learned $y = x \cdot w + b$. This is basically a single **Neuron** (or linear layer). It takes inputs, multiplies them by weights, adds a bias, and outputs a number.

A **Neural Network** is simply stacking these operations. We group neurons into **Layers**. 
- **Input Layer**: The raw data features.
- **Hidden Layers**: Layers in between input and output where the "magic" representations happen.
- **Output Layer**: The final prediction.

However, if you just stack $y_1 = x \cdot w_1 + b_1$ and $y_2 = y_1 \cdot w_2 + b_2$, the math collapses into a single linear equation. To learn complex, curved, real-world patterns, we *must* introduce **Activation Functions** (non-linearities) between the layers.

## 5. Intuition & Mathematical Foundation
**Activation Function (ReLU)**: The Rectified Linear Unit (ReLU) is the most popular activation function. 
$f(x) = \max(0, x)$
If the number is negative, it becomes 0. If it's positive, it stays the same. This simple kink in the graph is what gives Neural Networks the power to approximate *any* function (The Universal Approximation Theorem).

```
# Let's visualize ReLU
x_vals = torch.linspace(-5, 5, 100)
relu_vals = torch.relu(x_vals) # Built-in PyTorch activation

plt.plot(x_vals.numpy(), relu_vals.numpy())
plt.title("ReLU Activation Function")
plt.grid()
plt.show()
```

## 8. Simple Example: Raw Forward Pass
Let's build a Neural Network with 1 hidden layer without using `torch.nn` to understand exactly what happens under the hood.

```
# 1 Batch of 5 samples, each with 3 features
X = torch.randn(5, 3)

# --- Layer 1 (Hidden Layer) ---
# 3 input features -> 4 hidden neurons
W1 = torch.randn(3, 4)
b1 = torch.randn(4)

# Linear transformation
Z1 = X @ W1 + b1

# Activation
A1 = torch.relu(Z1)

# --- Layer 2 (Output Layer) ---
# 4 hidden neurons -> 1 output prediction
W2 = torch.randn(4, 1)
b2 = torch.randn(1)

# Linear transformation
output = A1 @ W2 + b2

print("Final Output Predictions:\n", output)
```

## 9. Code Walkthrough
1. `X @ W1 + b1`: Matrix multiplication creates connections from every input feature to every hidden neuron.
2. `torch.relu(Z1)`: Applies the non-linearity. Without this, the network would just be a fancy linear regression.
3. `A1 @ W2 + b2`: Takes the activated signals and maps them to the final output.

## 13. Debugging Challenge
Why does this forward pass crash? How do you fix it?

```
inputs = torch.randn(16, 20)
weights = torch.randn(64, 20)
bias = torch.randn(64)

# output = inputs @ weights + bias # Uncomment to see error
```

**Solution:** In `inputs @ weights`, the inner dimensions must match. `inputs` is `[16, 20]` and `weights` is `[64, 20]`. PyTorch's `nn.Linear` expects the weight matrix to be shaped `[out_features, in_features]`, so it internally transposes the weights. 

Fix it by either defining `weights = torch.randn(20, 64)` OR doing `inputs @ weights.T`.

## 17. Interview Questions
1. **What happens if you remove all activation functions from a 50-layer deep neural network?**
   *Answer*: The entire 50-layer network mathematically collapses into a single linear transformation (equivalent to logistic/linear regression). It loses the ability to learn non-linear patterns.
2. **What are the Parameters of a neural network?**
   *Answer*: The weights and biases. These are the values the network learns and updates during training.

## 19. Day Summary
- Neural Networks are stacked sequences of Linear Transformations (`X @ W + b`).
- Activation functions (`ReLU`) are placed between layers to inject non-linearity.
- The shapes of your weight matrices strictly dictate the architecture of your network.
