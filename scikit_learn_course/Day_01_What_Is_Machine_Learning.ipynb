## 3. Concept
**Machine Learning (ML)** is a subset of AI where systems learn from data to make predictions or decisions without being explicitly programmed.

Types of ML:
- **Supervised Learning**: The dataset has labels (features `X` and target `y`). The goal is to learn a mapping from `X` to `y`.
  - *Regression*: Predicting a continuous value (e.g., house price).
  - *Classification*: Predicting a categorical value (e.g., spam vs. not spam).
- **Unsupervised Learning**: The dataset has no labels. The goal is to find hidden structures.
  - *Clustering*: Grouping similar data points together.
- **Semi-supervised Learning**: A mix of a small amount of labeled data and a large amount of unlabeled data.


## 4. Why Does This Matter?
Traditional programming requires you to write the rules: `Data + Rules -> Answers`.

Machine learning flips this paradigm: `Data + Answers -> Rules`.

Without ML, solving complex problems like image recognition, language translation, or predicting customer churn would require an impossible number of `if/else` statements.


## 5. Intuition
Imagine teaching a child to recognize an apple.
You don't write rules like `if color == red and shape == round`. Instead, you show them dozens of apples and say "This is an apple." Over time, they learn the pattern.

In ML:
- The examples you show are the **Samples**.
- The characteristics (color, shape, weight) are the **Features (X)**.
- The label ("apple") is the **Target (y)**.
- The learning process is **Training**.
- When they correctly identify a new apple, that's **Inference**.


## 6. Mathematical Foundation
At its core, supervised machine learning is about approximating a function $f$ such that:

$$y \approx f(X)$$

Where:
- $X$ is an $n \times p$ matrix ($n$ samples, $p$ features).
- $y$ is an $n$-dimensional vector of targets.
- The algorithm estimates $\hat{f}$ to minimize some error between the predicted $\hat{y}$ and true $y$.

## 8. Simple Example
Let's generate a synthetic dataset and fit a simple classification model.


```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt

# 1. Create data (Features X, Target y)
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0, random_state=42)

# 2. Initialize the model
model = LogisticRegression()

# 3. Train the model
model.fit(X, y)

# 4. Inference (Predict)
predictions = model.predict(X[:5])
print('First 5 true targets:', y[:5])
print('First 5 predictions: ', predictions)
```

## 9. Code Walkthrough
- `make_classification()`: Generates a synthetic dataset for classification.
- `X`: A 2D array of shape `(100, 2)` representing 100 samples and 2 features.
- `y`: A 1D array of shape `(100,)` representing the class labels (0 or 1).
- `LogisticRegression()`: Instantiates a classification algorithm.
- `model.fit(X, y)`: The model looks at X and y and learns the relationship (Training).
- `model.predict(X[:5])`: The model predicts the target for the first 5 samples (Inference).


## 10. Experiment

## 17. Common Mistakes
- **Confusing Features and Targets**: `X` is always your input data (features), `y` is the outcome you want to predict (target).
- **Shape Errors**: Passing 1D arrays to models. `X` must be 2D `(n_samples, n_features)`.
- **Import Errors**: Forgetting which module a class belongs to (e.g., `sklearn.linear_model`).
