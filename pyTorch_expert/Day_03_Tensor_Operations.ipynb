## 3. Concept Explanation
Machine Learning involves applying mathematical operations to huge matrices of data. PyTorch provides hundreds of operations to manipulate these tensors.

**Broadcasting** is a set of rules PyTorch uses to perform operations on tensors of different sizes. If one tensor is `[3, 3]` and you add a scalar `1` to it, PyTorch *broadcasts* the `1` across all 9 elements without explicitly duplicating it in memory.

## 4. Why This Matters
Neural networks are simply a long sequence of tensor multiplications and additions, followed by activation functions. If you don't understand how `torch.matmul` differs from the `*` operator, you cannot build a neural network.

## 5. Intuition
**Element-wise (`*`)**: Multiply each element in tensor A with the corresponding element in tensor B at the exact same position.
**Matrix Multiplication (`@` or `torch.matmul`)**: Multiply rows by columns (Standard linear algebra multiplication).

```
# 8. Simple Example: Arithmetic & Matrix Multiplication
tensor_A = torch.tensor([[1, 2], [3, 4]])
tensor_B = torch.tensor([[1, 1], [1, 1]])

print("Addition:\n", tensor_A + tensor_B)
print("\nElement-wise Mult:\n", tensor_A * tensor_B)
print("\nMatrix Mult:\n", tensor_A @ tensor_B) # Equivalent to torch.matmul(tensor_A, tensor_B)
```

## 9. Code Walkthrough: Reductions
Reduction operations reduce the number of dimensions in a tensor by computing aggregates.

```
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print("Original:\n", x)

print("\nSum all:", x.sum())
print("Sum over columns (dim=0):", x.sum(dim=0)) # Sum vertically
print("Sum over rows (dim=1):", x.sum(dim=1))    # Sum horizontally

print("\nArgmax (index of max value):", x.argmax())
```

## 10. Experiment: Broadcasting
Let's see broadcasting in action.

```
matrix = torch.zeros(3, 3)
vector = torch.tensor([1, 2, 3])

# The vector [1,2,3] is broadcasted across all rows of the matrix!
result = matrix + vector
print("Broadcast Result:\n", result)
```

## 13. Debugging Challenge
Why does this matrix multiplication fail?

```
tensor_1 = torch.rand(3, 2)
tensor_2 = torch.rand(3, 2)

# result = torch.matmul(tensor_1, tensor_2) # Uncomment to see error
```

**Solution:** In matrix multiplication, the inner dimensions must match! `[3, 2] @ [3, 2]` is invalid because `2 != 3`. 
Fix it by transposing the second tensor: `tensor_1 @ tensor_2.T` which changes the operation to `[3, 2] @ [2, 3] = [3, 3]`.

## 17. Interview Questions
1. **What is Broadcasting in PyTorch?**
   *Answer*: It is how PyTorch handles operations between tensors of different shapes. The smaller tensor is virtually expanded to match the larger tensor without actually copying data.
2. **Explain `argmax`.**
   *Answer*: It returns the *index* of the maximum value in a tensor, rather than the value itself. It is commonly used to find the predicted class in classification problems.

## 19. Day Summary
- Use `+`, `-`, `*`, `/` for element-wise arithmetic.
- Use `@` or `torch.matmul` for matrix multiplication (inner dimensions must match).
- Use `dim` argument in reduction operations (`sum`, `mean`, `max`) to collapse a specific dimension.
- Broadcasting allows arithmetic on tensors of different, but compatible shapes.
